# 三、queue 核心常用操作（共6个，只记这一套）

## 逐函数详细演示（零基础必看）
```cpp
// 创建任意底层的队列（操作完全一样）
queue<int> q;

// 1. push(x)：队尾入队（加元素）
q.push(10);   // 队列：[10]
q.push(20);   // 队列：[10, 20]
q.push(30);   // 队列：[10, 20, 30] （10队头，30队尾）

// 2. front()：取队头元素
cout << "队头元素：" << q.front() << endl;  // 输出 10

// 3. back()：取队尾元素
cout << "队尾元素：" << q.back() << endl;   // 输出 30

// 4. pop()：删除队头元素（只删，不返回值）
q.pop();  // 删除10，队列变成：[20, 30]
cout << "新队头：" << q.front() << endl;    // 输出 20

// 5. empty()：判断队列是否为空
if (q.empty()) cout << "队列为空！" << endl;
else cout << "队列不为空！" << endl;

// 6. size()：获取元素个数
cout << "元素个数：" << q.size() << endl;  // 输出 2
```

# 四、完整可运行测试代码（直接复制运行）
```cpp
#include <iostream>
#include <queue>
using namespace std;

int main() {
    // 创建队列
    queue<int> q;

    // 入队（队尾添加）
    q.push(5);
    q.push(15);
    q.push(25);

    // 查看队头、队尾
    cout << "队头：" << q.front() << endl;   // 5
    cout << "队尾：" << q.back() << endl;    // 25

    // 出队（删除队头）
    q.pop();
    cout << "出队后队头：" << q.front() << endl;  // 15

    // 大小 + 判空
    cout << "元素个数：" << q.size() << endl;    // 2
    cout << "是否为空：" << q.empty() << endl;   // 0

    // 清空队列（手动循环出队）
    while (!q.empty()) {
        q.pop();
    }
    cout << "清空后是否为空：" << q.empty() << endl; // 1

    return 0;
}
```

# 三、逐函数详细演示（零基础必看）
```cpp
// 创建一个空队列（任意底层都可以，用法完全一致）
queue<int> q;

// 1. push(x)：队尾入队（添加元素）
q.push(10);   // 队列：队头 [10] 队尾
q.push(20);   // 队列：[10, 20]
q.push(30);   // 队列：[10, 20, 30]

// 2. front()：查看队头元素（最早入队的元素）
cout << "队头元素：" << q.front() << endl;  // 输出 10

// 3. back()：查看队尾元素（最晚入队的元素）
cout << "队尾元素：" << q.back() << endl;   // 输出 30

// 4. pop()：删除队头元素（只删除，不返回值）
q.pop();  // 删除 10，队列变为 [20, 30]
cout << "新队头：" << q.front() << endl;    // 输出 20

// 5. empty()：判断队列是否为空
if (q.empty()) {
    cout << "队列为空！" << endl;
} else {
    cout << "队列不为空！" << endl;
}

// 6. size()：获取队列元素个数
cout << "元素总数：" << q.size() << endl;  // 输出 2
```
