# 一、双端队列 → ✅ STL 官方容器：`std::deque`
这是**唯一有原生STL实现**的结构，也是你问的四种里**最常用、必须掌握**的！
## 1. 核心特性
双端队列 = **头尾都能插入、删除**的队列（普通队列只能队尾入、队头出）
- 支持随机访问 `[]`
- 头尾插删效率极高
- 是 `stack` 和 `queue` 的**默认底层容器**

## 2. 基础用法（零基础直接抄）
```cpp
#include <deque>  // 头文件
using namespace std;

// 1. 创建空双端队列
deque<int> dq;

// 2. 核心操作（头尾自由操作）
dq.push_back(10);   // 尾部入
dq.push_front(20);  // 头部入（普通queue没有！）
dq.pop_back();      // 尾部出
dq.pop_front();     // 头部出
dq.front();         // 取头
dq.back();          // 取尾
dq.size();          // 大小
dq.empty();         // 判空
```

## 3. 总结
**双端队列 = STL 亲儿子 `deque`，直接用！**

# 二、双栈 → ❌ 无原生容器，用 **两个 stack 组合**
## 1. 定义
双栈 = **两个栈配合**实现特定功能（最经典：用双栈实现队列）
## 2. STL 实现
没有专门的 `std::double_stack`，**直接定义两个 `stack`** 即可：
```cpp
stack<int> st1;
stack<int> st2;
// 两个栈互相倒数据，就能实现队列功能
```
## 3. 总结
双栈是**逻辑结构**，不是新容器，用你已经学的 `stack` 拼起来就行。

# 三、超队列 → ❌ 无原生容器，**直接用 deque 实现**
## 1. 定义
超队列 = **增强版队列**：
队头/队尾 **都能入队、都能出队**（功能比普通 queue 强）
## 2. STL 实现
超队列的所有功能，**`deque` 天生全部支持**，直接用 `deque` 就是超队列：
```cpp
deque<int> 超队列;
超队列.push_front(); // 队头入
超队列.push_back();  // 队尾入
超队列.pop_front();  // 队头出
超队列.pop_back();   // 队尾出
```

# 四、超栈 → ❌ 无原生容器，**直接用 deque 实现**
## 1. 定义
超栈 = **增强版栈**：
栈顶/栈底 **都能压入、都能弹出**（功能比普通 stack 强）
## 2. STL 实现
超栈的所有功能，**`deque` 天生全部支持**，直接用 `deque` 就是超栈：
```cpp
deque<int> 超栈;
超栈.push_back();  // 栈顶压入
超栈.pop_back();   // 栈顶弹出
超栈.push_front(); // 栈底压入
超栈.pop_front();  // 栈底弹出
```

# 零基础完全上手 `std::deque`（双端队列）超全指南

## 🔥 核心前置知识
1. **deque = 双端队列**，头尾都能自由增删元素
2. 支持 `[]` 随机访问，**没有 list 不能下标访问的限制**
3. 头文件：`#include <deque>`

# 一、deque 的创建与初始化（和 vector 完全一样）
共 5 种最常用创建方式，零基础直接抄：
```cpp
#include <iostream>
#include <deque>
using namespace std;

int main() {
    // 1. 创建空双端队列（最常用）
    deque<int> dq1;  

    // 2. 初始化列表（直接赋值）
    deque<int> dq2 = {1, 2, 3, 4, 5};  
    deque<int> dq3{10, 20, 30};        

    // 3. 创建 n 个相同值的元素
    deque<int> dq4(5, 10);  // 5个10：10 10 10 10 10

    // 4. 拷贝另一个 deque
    deque<int> dq5(dq2);  

    // 5. 用数组初始化
    int arr[] = {100, 200, 300};
    deque<int> dq6(arr, arr + 3);  

    return 0;
}
```

---

# 二、deque 访问元素（支持下标！比 list 强太多）
**和 vector 访问方式完全一致**，这是双端队列的核心优势：
```cpp
deque<int> dq = {1,2,3,4,5};

// 1. 下标访问 []（最常用，和数组/vector一样）
cout << dq[0];   // 输出 1
cout << dq[2];   // 输出 3

// 2. at() 访问（带越界检查，更安全）
cout << dq.at(1); // 输出 2

// 3. 获取头部元素
cout << dq.front(); // 输出 1

// 4. 获取尾部元素
cout << dq.back();  // 输出 5
```

---

# 三、deque 核心操作（⭐ 独有的：头部插删）
这是 **deque 区别于 vector 的核心功能**：
vector 头部插删极慢，deque 头尾插删都超快！

| 函数 | 功能 | 适用场景 |
|------|------|----------|
| `push_back(x)` | 尾部插入元素 | 和 vector 一致 |
| `pop_back()` | 删除尾部元素 | 和 vector 一致 |
| `push_front(x)` | **头部插入元素** | deque 独有核心功能 |
| `pop_front()` | **删除头部元素** | deque 独有核心功能 |

### 代码演示
```cpp
deque<int> dq;

// 尾部操作（vector 同款）
dq.push_back(10);
dq.push_back(20);  // 队列：[10,20]

// 头部操作（vector 没有！）
dq.push_front(5);  // 头部插入 → [5,10,20]
dq.pop_front();    // 删除头部 → [10,20]

// 尾部删除
dq.pop_back();     // 删除尾部 → [10]
```

# 四、中间插入/删除（和 vector 用法一致）
deque 支持在指定位置插入/删除，用法和 vector 完全相同：
```cpp
deque<int> dq = {1,3,4};

// 1. 指定位置插入单个元素（迭代器位置前插入）
auto it = dq.begin() + 1; // 指向 3
dq.insert(it, 2);         // 插入后：[1,2,3,4]

// 2. 删除指定位置元素
dq.erase(dq.begin() + 1); // 删除后：[1,3,4]
```

# 五、容量 / 大小操作（和 vector 完全一致）
```cpp
deque<int> dq = {1,2,3};

// 1. 获取元素个数
cout << dq.size(); // 输出 3

// 2. 判断是否为空
cout << dq.empty(); // 0（false）

// 3. 调整大小（保留原有元素，新增补0）
dq.resize(5); 

// 4. 预留空间（优化性能）
dq.reserve(100);
```

# 六、遍历 deque（3 种方式，和 vector 完全一样）
### 方式1：下标遍历（最简单，首选）
```cpp
for(int i=0; i<dq.size(); i++){
    cout << dq[i] << " ";
}
```

### 方式2：范围 for 循环
```cpp
for(int x : dq){
    cout << x << " ";
}
```

### 方式3：迭代器遍历
```cpp
for(auto it = dq.begin(); it != dq.end(); ++it){
    cout << *it << " ";
}
```

# 八、完整可运行测试代码（直接复制运行）
```cpp
#include <iostream>
#include <deque>
using namespace std;

int main() {
    // 1. 创建
    deque<int> dq;

    // 2. 头尾插入
    dq.push_back(20);
    dq.push_back(30);
    dq.push_front(10);  // 头部插入 → [10,20,30]

    // 3. 访问
    cout << "头部：" << dq.front() << endl;
    cout << "尾部：" << dq.back() << endl;
    cout << "下标1：" << dq[1] << endl;

    // 4. 遍历
    cout << "遍历：";
    for (int x : dq) cout << x << " ";  // 10 20 30
    cout << endl;

    // 5. 头尾删除
    dq.pop_front();
    dq.pop_back();
    cout << "删除后：";
    for (int x : dq) cout << x << " ";  // 20

    return 0;
}
```
