# 一、vector 的创建（初始化）【最基础】
一共 **6 种最常用创建方式**，覆盖所有场景：
```cpp
#include <iostream>
#include <vector>
using namespace std;

int main() {
    // 1. 创建空 vector（最常用）
    vector<int> v1;  

    // 2. 创建指定大小的 vector（初始值默认为0）
    vector<int> v2(5);  // 大小5，元素：0 0 0 0 0

    // 3. 创建指定大小 + 统一初始值
    vector<int> v3(5, 10);  // 大小5，元素：10 10 10 10 10

    // 4. 直接初始化（C++11）
    vector<int> v4 = {1,2,3,4,5};  
    vector<int> v5{1,2,3}; // 简写

    // 5. 拷贝另一个vector
    vector<int> v6(v5);  // 和v5完全一样

    // 6. 用数组初始化
    int arr[] = {10,20,30};
    vector<int> v7(arr, arr+3); // 数组转vector
    return 0;
}
```

# 二、访问元素（和数组一样用）
`vector` 支持**随机访问**（O(1) 时间，顺序表核心特性）
```cpp
vector<int> v = {10,20,30,40};

// 1. 下标访问 [ ] （最常用，和数组一模一样）
cout << v[0];  // 输出10
cout << v[2];  // 输出30

// 2. at() 访问（带越界检查，更安全）
cout << v.at(1); // 输出20

// 3. 获取第一个元素
cout << v.front(); // 10

// 4. 获取最后一个元素
cout << v.back();  // 40
```

# 三、增加元素（顺序表插入操作）
分 **尾部插入**（最快）和 **指定位置插入**
```cpp
vector<int> v;

// 1. 尾部插入元素 push_back（最常用！）
v.push_back(1);
v.push_back(2);
v.push_back(3); 
// 现在 v：[1,2,3]

// 2. 指定位置插入 insert
// 格式：v.insert(迭代器位置, 值)
v.insert(v.begin(), 0);    // 开头插入0 → [0,1,2,3]
v.insert(v.begin()+2, 99); // 下标2插入99 → [0,1,99,2,3]
```
✅ 重点：`v.begin()` 是**首位置迭代器**，`v.begin()+n` = 下标 `n`

# 四、删除元素（顺序表删除操作）
```cpp
vector<int> v = {1,2,3,4,5};

// 1. 删除最后一个元素 pop_back（最快）
v.pop_back(); // 删5 → [1,2,3,4]

// 2. 删除指定位置 erase
v.erase(v.begin()); // 删开头 → [2,3,4]
v.erase(v.begin()+1); // 删下标1 → [2,4]

// 3. 清空所有元素 clear
v.clear(); // 空vector
```

# 二、resize 操作详解（最关键！）
## 核心结论
1. **resize 不会破坏、修改原有元素！** 原有数据完全保留
2. **可以自定义新增元素的值！不是只能初始化为 0！**
3. resize = 改变 **vector 的实际元素个数（size）**

## 分两种情况
### 1. 变大（resize(更大的数)）
- 原有元素不动
- 新增的元素可以**指定值**
```cpp
vector<int> v = {1,2,3}; 
v.resize(5, 10); // 变大到5个元素，新增的2个设为10
// 结果：[1,2,3,10,10] ✅ 原有元素不变，新增可自定义
```
如果不指定值，默认补 `0`：
```cpp
v.resize(5); 
// 结果：[1,2,3,0,0]
```

### 2. 变小（resize(更小的数)）
- 保留前面的元素
- 删掉后面多余的元素
```cpp
vector<int> v = {1,2,3,4,5};
v.resize(3);
// 结果：[1,2,3]
```

# 三、reserve 到底有什么用？为什么要预留空间？
## 核心结论
1. reserve **只分配空间，不增加元素**（size 不变，不创建元素）
2. reserve = **提前占坑**，避免 vector 自动频繁扩容
3. 自动扩容很慢、很耗性能！这就是 reserve 的意义

## 通俗解释
vector 默认**满了才扩容**（比如扩到原来的 2 倍），频繁插入会频繁扩容：
- 每次扩容都要**重新申请内存 + 拷贝所有旧元素** → 巨慢
- reserve 提前一次性开好足够大的空间，**只拷贝一次**

## 代码示例
```cpp
vector<int> v;
v.reserve(100); // 预分配 100 个位置，不创建元素
// 接下来插入 100 个元素，不会触发任何扩容！
```

## resize vs reserve（必考/必懂区别）
| 操作 | 改变 size（元素个数） | 改变 capacity（容量） | 创建元素 | 用途 |
|------|----------------------|----------------------|----------|------|
| resize | ✅ 是 | ✅ 是 | ✅ 是 | 调整实际元素数量 |
| reserve | ❌ 否 | ✅ 是 | ❌ 否 | 提前预分配空间，优化速度 |

# 一、`vector<int> v7(arr, arr+3);` 格式 + 含义
## 1. 标准格式
```cpp
vector<类型> 变量名(起始地址, 结束地址);
```
这是 **「区间初始化」**（也叫迭代器初始化），专门用来**把一段连续的数据（数组、另一个vector）拷贝进新vector**。

## 2. 关键规则（必背）
- **左闭右开**：`[起始地址, 结束地址)`
- 包含**起始地址**的元素，**不包含**结束地址的元素
- `arr` = 数组首地址（第 0 个元素）
- `arr+3` = 数组往后偏移 3 个位置（第 3 个元素的地址，不包含）

## 3. 完整示例
```cpp
int arr[] = {10, 20, 30, 40}; 
// arr = 10的地址，arr+1=20，arr+2=30，arr+3=40，arr+4=数组末尾
vector<int> v7(arr, arr+3); 
```
✅ **含义**：拷贝数组中 **第0、1、2号元素**（共3个）
✅ **结果**：`v7 = {10, 20, 30}`

## 4. 一句话总结
`(arr, arr+n)` = 把数组的 **前 n 个元素** 放进 vector。

# 二、vector 默认初始预留多少空间？什么时候扩容？
## 核心结论（竞赛/开发通用）
1. **空 vector 初始预留空间 = 0**
   `vector<int> v;` → `size=0`，`capacity=0`（没有任何预留空间）
2. **扩容规则（倍数扩容）**
   - **GCC 编译器（竞赛/Linux 最常用）**：**2 倍扩容**
     `0 → 1 → 2 → 4 → 8 → 16 → 32 → 64...`
   - **VS 编译器（Windows）**：**1.5 倍扩容**
3. **触发时机**：插入元素时，`size == capacity`（满了），立刻扩容

## 直观演示（GCC 环境）
```cpp
vector<int> v;
v.push_back(1);  // capacity: 1
v.push_back(2);  // 满了→扩容2倍 → capacity:2
v.push_back(3);  // 满了→扩容2倍 → capacity:4
v.push_back(4);  // 不扩容
v.push_back(5);  // 满了→扩容2倍 → capacity:8
```

## 补充
没有固定的“初始预留大小”，完全靠**插入元素触发自动扩容**。

# 三、超级重要纠正：
## ❌ `emplace_back` **绝对不是批量添加**！
## ✅ 它和 `push_back` 一样，都是**添加【单个】元素**！

### 1. 你为什么会误解？
大概率是把 `emplace_back` 和 `insert` 搞混了：
- **批量添加**：用 `insert`
- **单个尾插**：`push_back` / `emplace_back`

### 2. 两者用法**完全一致**（都是加1个元素）
```cpp
vector<int> v;
// push_back：尾插一个元素
v.push_back(10);  

// emplace_back：也是尾插一个元素
v.emplace_back(20); 

// 最终结果：{10,20}
```

### 3. 唯一区别（底层，不用深究，知道即可）
- `push_back(值)`：先创建元素，再**拷贝**到 vector
- `emplace_back(值)`：**直接在 vector 内部创建元素**（效率更高一点）

### 4. 真正的「批量添加」怎么写？
```cpp
vector<int> v;
// 批量添加 3 个 5
v.insert(v.end(), 3, 5);
```

# 一、array 的创建与初始化（最基础，必背）

共 **5 种最常用初始化方式**：

```cpp
#include <iostream>
#include <array>  // array 头文件
using namespace std;

int main() {
    // 1. 创建空 array（默认初始化：int 类型为随机值，全局/静态为0）
    array<int, 5> a1;  

    // 2. 列表初始化（最常用，直接赋值）
    array<int, 5> a2 = {1, 2, 3, 4, 5}; 
    array<int, 5> a3{10, 20, 30};       // C++11 简写，剩余元素自动补0 → [10,20,30,0,0]

    // 3. 全部元素初始化为 0（极简写法）
    array<int, 5> a4 = {0};  

    // 4. 拷贝另一个 array
    array<int, 5> a5(a2);  // 和 a2 完全一样

    // 5. 全部元素赋统一值（用 fill 函数）
    array<int, 5> a6;
    a6.fill(100);  // 所有元素 = 100 → [100,100,100,100,100]

    return 0;
}
```

```cpp
for(auto it = a.begin(); it != a.end(); it++){
    cout << *it << " ";
}
```

什么条件下两个array才能用swap以及赋值呢？array可以用clear操作吗？

我来用**最直白、精准、带代码示例**的方式，一次性把你问的两个核心问题讲透，完全贴合 `std::array` 的规则！

# 一、两个 array 能赋值 / swap 的**唯一条件**
## 核心规则（死记硬背）
两个 `std::array` 想要 **互相赋值** 或 **调用 swap 交换**，必须满足：
### ✅ **元素类型完全相同 + 数组大小完全相同**
`array` 的完整类型是 `array<数据类型, 固定大小>`，**这两个参数必须一模一样**！

---

## 1. 合法示例（可以赋值 + swap）
```cpp
#include <array>
using namespace std;

// 两个都是：int类型 + 大小5 → 完全匹配
array<int, 5> a1 = {1,2,3,4,5};
array<int, 5> a2 = {10,20,30,40,50};

// ✅ 直接赋值（把a2所有元素拷贝给a1）
a1 = a2;  

// ✅ swap 交换（两个数组的元素全部互换）
a1.swap(a2); 
```

---

## 2. 非法示例（绝对报错）
只要**类型不同** 或 **大小不同**，都不能赋值/swap：
```cpp
array<int, 5> a;
array<int, 6> b;   // ❌ 大小不同 → 报错
array<double,5> c; // ❌ 类型不同 → 报错

a = b;    // 编译失败
a.swap(c);// 编译失败
```

---

## 简单总结
`array<int, 5>` 和 `array<int, 5>` 才能互相操作
**类型不一样不行，大小不一样不行**

---

# 二、array 可以用 clear() 操作吗？
## 终极答案：**绝对不可以！❌**
### 原因（根本原理）
1. `array` 是**静态固定大小**的数组，**大小一旦创建永远不能改变**
2. `clear()` 的作用是：**清空所有元素，让 size = 0**
3. `array` 的 `size()` 是固定的（创建时写死的数字），**永远不可能变成 0**

所以 **`array` 根本没有 clear() 这个函数！**

---

## 想要“清空/重置”array，用这个替代方法
用 **`fill(初始值)`** 把所有元素批量重置（等效清空）：
```cpp
array<int, 5> a = {1,2,3,4,5};

// ✅ 等效清空：把所有元素重置为 0
a.fill(0); 

// ✅ 也可以重置为其他值
a.fill(-1);
```

| 操作         | std::array (静态固定大小) | std::vector (动态可变) |
|--------------|---------------------------|------------------------|
| 赋值/swap 条件 | 类型+大小必须完全相同     | 只要类型相同即可       |
| clear()      | ❌ 无此函数               | ✅ 可以清空，size=0    |
| 大小可变     | ❌ 永远不变               | ✅ 随时扩容/缩容       |

# 四、零基础终极总结
1. **赋值 / swap**
   两个 `array` 的**数据类型 + 大小**必须**完全一样**，缺一不可。

2. **clear()**
   `array` **不能用 clear**，因为大小固定；
   想清空就用 **`a.fill(0)`** 重置所有元素。

3. 一句话记忆
   **array 是死大小，不能删、不能清、只能同类型同大小互换！**
