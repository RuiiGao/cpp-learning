# 二、stack 核心常用操作（共 5 个，死记这一套）
这 5 个操作是栈的全部功能，**不管底层是啥，用法完全相同**：

| 操作函数 | 功能 | 用法说明 | 代码示例 |
|----------|------|----------|----------|
| `push(x)` | **元素入栈** | 将元素 `x` 压入栈顶 | `st.push(10);` |
| `pop()` | **栈顶出栈** | **删除**栈顶元素，**无返回值** | `st.pop();` |
| `top()` | **取栈顶元素** | 获取栈顶元素的值，不删除 | `int x = st.top();` |
| `empty()` | **判断栈空** | 空返回 `true`，非空返回 `false` | `if(st.empty())` |
| `size()` | **获取元素个数** | 返回栈里有多少个元素 | `cout << st.size();` |

---

## 逐函数详细演示（零基础必看）
```cpp
// 创建一个空栈（任意底层都可以）
stack<int> st;

// 1. push(x)：入栈（往栈顶加元素）
st.push(10);   // 栈：[10]
st.push(20);   // 栈：[10, 20]
st.push(30);   // 栈：[10, 20, 30]  （30是栈顶）

// 2. top()：获取栈顶元素（只能看栈顶！）
cout << "栈顶元素：" << st.top() << endl;  // 输出 30

// 3. pop()：删除栈顶元素（只删，不返回值）
st.pop();  // 删除30，栈变成：[10, 20]
cout << "新栈顶：" << st.top() << endl;    // 输出 20

// 4. empty()：判断栈是否为空
if (st.empty()) {
    cout << "栈为空！" << endl;
} else {
    cout << "栈不为空！" << endl;
}

// 5. size()：获取栈中元素个数
cout << "栈的元素个数：" << st.size() << endl;  // 输出 2
```

```cpp
#include <iostream>
#include <stack>
using namespace std;

int main() {
    // 创建栈
    stack<int> st;

    // 入栈
    st.push(5);
    st.push(15);
    st.push(25);

    // 查看栈顶
    cout << "栈顶：" << st.top() << endl;  // 25

    // 出栈
    st.pop();
    cout << "出栈后栈顶：" << st.top() << endl;  // 15

    // 大小 + 判空
    cout << "元素个数：" << st.size() << endl;  // 2
    cout << "是否为空：" << st.empty() << endl; // 0（false）

    // 清空栈（手动循环出栈）
    while (!st.empty()) {
        st.pop();
    }
    cout << "清空后是否为空：" << st.empty() << endl; // 1（true）

    return 0;
}
```
